# Fashion Recommendation System - Experimentation Notebook

This notebook is for experimenting with the different layers of the fashion recommendation system.

## Setup

```python
import sys
sys.path.insert(0, '..')

from src.core.models import (
    Garment, GarmentAttributes, GarmentCategory,
    ColorInfo, PatternInfo, FormalityLevel, Season,
    UserContext, WeatherContext, Occasion
)
from src.layer2_style import (
    StyleIntelligenceModel, ColorHarmonyAnalyzer,
    SilhouetteAnalyzer, FormalityMatcher
)
from src.layer3_context import ContextEngine, OccasionAnalyzer
```

## Layer 2: Style Intelligence Experiments

```python
# Create sample garments
white_tshirt = Garment(
    id="white_tshirt",
    attributes=GarmentAttributes(
        category=GarmentCategory.TOP,
        subcategory="t-shirt",
        color=ColorInfo(primary="white", hex_codes=["#FFFFFF"]),
        pattern=PatternInfo(type="solid"),
        style_tags=["casual", "minimalist"],
        formality_level=FormalityLevel.CASUAL,
        season_suitable=[Season.SPRING, Season.SUMMER],
        silhouette="relaxed",
        fit="regular"
    )
)

blue_jeans = Garment(
    id="blue_jeans",
    attributes=GarmentAttributes(
        category=GarmentCategory.BOTTOM,
        subcategory="jeans",
        color=ColorInfo(primary="blue", hex_codes=["#1E40AF"]),
        pattern=PatternInfo(type="solid"),
        style_tags=["casual", "classic"],
        formality_level=FormalityLevel.CASUAL,
        season_suitable=[Season.SPRING, Season.FALL, Season.WINTER],
        silhouette="straight",
        fit="slim"
    )
)

print(f"Created: {white_tshirt.id} and {blue_jeans.id}")
```

```python
# Test Color Harmony
color_analyzer = ColorHarmonyAnalyzer()

harmony_score = color_analyzer.analyze_outfit_colors([white_tshirt, blue_jeans])
print(f"Color harmony score: {harmony_score:.2f}")

# Get complementary colors
complements = color_analyzer.get_complementary_colors("navy")
print(f"Colors that go with navy: {complements}")
```

```python
# Test Formality Matching
formality_matcher = FormalityMatcher()

formality_score = formality_matcher.check_formality_consistency([white_tshirt, blue_jeans])
print(f"Formality consistency: {formality_score:.2f}")

# Get compatible formalities
compatible = formality_matcher.get_compatible_formalities(FormalityLevel.SMART_CASUAL)
print(f"Compatible with smart-casual: {[f.value for f in compatible]}")
```

```python
# Test Full Style Model
style_model = StyleIntelligenceModel()

# Score the outfit
import asyncio

async def score_outfit():
    scores = await style_model.score_outfit([white_tshirt, blue_jeans])
    return scores

scores = asyncio.get_event_loop().run_until_complete(score_outfit())
print("Outfit Scores:")
print(f"  Overall: {scores['overall_score']:.2f}")
for key, value in scores['breakdown'].items():
    print(f"  {key}: {value:.2f}")
```

## Layer 3: Context Engine Experiments

```python
# Test Occasion Analyzer
occasion_analyzer = OccasionAnalyzer()

for occasion in [Occasion.CASUAL, Occasion.BUSINESS, Occasion.FORMAL]:
    score = occasion_analyzer.score_for_occasion([white_tshirt, blue_jeans], occasion)
    print(f"{occasion.value}: {score:.2f}")
```
